```python
data = {
    'Name': ['Mahesh','Ram','Harika','Harini','Shankar','Sravani','Deekshith','Chanikhya'],
    'Age': [27,21,19,20,21,30,10,15],
    'City': ['Hyderabad','Vizag','Vizag','Hyd','Vizag','Hyderabad','Bangalore','Chennai'],
    'Salary': [20000,30000,40000,45000,50000,30000,35000,25000]}

df =pd.DataFrame(data,index=[101,102,103,104,105,106,107,108])
print(df)
```

```python
print(df['Age'].sum())
print(df['Age'].max())
print(df['Age'].min())
print(df['Age'].mean())
print(df['Age'].var())
print(df['Age'].std())
```

```python
print(df['Salary'].sum())
print(df['Salary'].max())
print(df['Salary'].min())
print(df['Salary'].mean())
print(df['Salary'].var())
print(df['Salary'].std())
```

```python
df['Bonus']=df['Salary']*0.10
print(df)
```

```python
df['new_salary']=df['Salary']+df['Bonus']
print(df)
```

```python
df['Age']=df['Age']+1
print(df)
```

```python
df['new_salary']=df['new_salary']+500
print(df)
```

```python
df.loc[df['Name'] =='Ram','City'] = 'Hyderabad'
print(df)
```

```python
df.loc[df['City'] == 'Bangalore','City'] = 'Hyderabad'
print(df)
```

```python
df.loc[df['City']=='Hyderabad','Salary']=df["Salary"]*1.05
print(df)
```

```python
df.loc[df['Salary']<=45000,'Salary']=60000
print(df)
```

```python
df.drop('Bonus',axis=1,inplace=True)
print(df)
```

```python
df.drop(columns='new_salary',inplace=True)
print(df)
```

Handling Duplicates


```python
import pandas as pd
data = {
    "EmpID":[101,102,103,104,105,106,107,108,109,110,110],

    "Name":["Ravi","PRIYA","Arun","sneha","Kiran","ANJALI","Vikram","meena","RAHUL","Sneha","Sneha"],

    "City":["Hyderabad","hyderabad","CHENNAI","Chennai","Bangalore","BANGALORE","Mumbai","mumbai","Delhi","Hyderabad","Hyderabad"],

    "Salary":[30000,40000,35000,45000,32000,50000,38000,42000,36000,45000,45000]

}

df=pd.DataFrame(data)

print(df)
```

```python
df.duplicated()
```

```python
df[df.duplicated()]
```

```python
df[df.duplicated()]
```

```python
df.duplicated().sum()
```

```python
d=df[df.duplicated()]
total = d[["Name"]].sum()
print(total)
```

```python
duplicate_rows = df[df.duplicated(keep=False)]
print(duplicate_rows)
```

```python
df["EmpID"].duplicated()
```

```python
d=df[df.duplicated()]
total = d[["Name"]].sum()
duplicate_rows = df[df.duplicated(keep=False)]
print(duplicate_rows)
f=df["EmpID"].duplicated()
print(f)
```

```python
df.drop_duplicates()
```

```python
df.drop_duplicates(keep="last")
```

```python
df.drop_duplicates(keep="first")
```

```python
df.drop_duplicates(keep=False)
```

```python
duplicate_all = df[df.duplicated(keep=False)]
display(duplicate_all)
```

```python
df.duplicated().sum()
```

```python
df[df.duplicated(keep=False)]
```

```python
amount=df[df.duplicated()]
print(amount["Salary"])
```
